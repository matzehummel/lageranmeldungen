```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient
import pandas as pd
from datetime import datetime

load_dotenv()
```

```python
mongodb_connection_string = os.environ.get('MONGODB_CONNECTION_STRING')
mongoClient = MongoClient(mongodb_connection_string)
db = mongoClient.lageranmeldungen
print("Connected to database")

# Set up the collection name
registrations_kinderlager = db['registrations']
registrations_xxl = db['registrations_xxl']

# Get all entries in the collection and convert to a list of dictionaries
data = list(registrations_kinderlager.find())
data_xxl = list(registrations_xxl.find())
```

```python
# Create a DataFrame from the list of dictionaries
df_kinderlager = pd.DataFrame(data)
df_kinderlager = df_kinderlager.drop('_id', 1)

df_xxl = pd.DataFrame(data_xxl)
df_xxl = df_xxl.drop('_id', 1)
#df_kinderlager["birthday"] = df_kinderlager["birthday"].apply(lambda x: datetime.strptime(x, '%Y-%m-%d').date())
#print(df_kinderlager.columns)
#df_kinderlager
```

```python
df_kinderlager["city"] = df_kinderlager["city"].str.strip()
```

```python
df_kinderlager
```

```python
m = df_kinderlager.query('sexe == "m"')
md = df_kinderlager.query('sexe == "m" and (city == "Doeggingen" or memberOf == "Ministranten Doeggingen")')
md["group"] = "md"
m1 = df_kinderlager.query('sexe == "m" and birthday > "2012-07-29" and city != "Doeggingen"')
m1["group"] = "m1"
m2 = df_kinderlager.query('sexe == "m" and birthday < "2012-07-30" and birthday > "2011-07-29" and city != "Doeggingen"')
m2["group"] = "m2"
m3 = df_kinderlager.query('sexe == "m" and birthday < "2011-07-30" and birthday > "2010-07-29" and city != "Doeggingen"')
m3["group"] = "m3"
m4 = df_kinderlager.query('sexe == "m" and birthday < "2010-07-30" and city != "Doeggingen"')
m4["group"] = "m4"

w = df_kinderlager.query('sexe == "w"')
wd = df_kinderlager.query('sexe == "w" and (city == "Doeggingen" or memberOf == "Ministranten Doeggingen")')
wd["group"] = "wd"
w1 = df_kinderlager.query('sexe == "w" and birthday > "2012-07-29" and city != "Doeggingen"')
w1["group"] = "w1"
w2 = df_kinderlager.query('sexe == "w" and birthday < "2012-07-30" and birthday > "2011-07-29" and city != "Doeggingen"')
w2["group"] = "w2"
w3 = df_kinderlager.query('sexe == "w" and birthday < "2011-07-30" and birthday > "2010-07-29" and city != "Doeggingen"')
w3["group"] = "w3"
w4 = df_kinderlager.query('sexe == "w" and birthday < "2010-07-30" and city != "Doeggingen"')
w4["group"] = "w4"

print("Kinder insgesamt:", len(df_kinderlager))
print("-------------------------------")
print("Jungs insgesamt:", len(m))
print("Dögginger Jungs:", len(md))
print("M1 (Jari und Luis):", len(m1))
print("M2 (Loris):", len(m2))
print("M3 (Ringo):", len(m3))
print("M4 (Lukas und March):", len(m4))
print("-------------------------------")
print("Mädels insgesamt:", len(w))
print("Dögginger Mädels:", len(wd))
print("W1 (Maria und Maja):", len(w1))
print("W2 (Lina und Vanessa):", len(w2))
print("W3 (Evelyn und Sophia):", len(w3))
print("W4 (Emilie und Sarah):", len(w4))
```
