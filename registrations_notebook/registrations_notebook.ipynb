```python
import os
from dotenv import load_dotenv
from pymongo import MongoClient
import pandas as pd
from datetime import datetime

load_dotenv(dotenv_path='../backend/.env')
```

```python
mongodb_connection_string = os.environ.get('MONGODB_CONNECTION_STRING')
mongoClient = MongoClient(mongodb_connection_string)
db = mongoClient.lageranmeldungen2026
print("Connected to database")

# Set up the collection name
registrations_kinderlager = db['kinderlager']
registrations_xxl = db['xxl']

# Get all entries in the collection and convert to a list of dictionaries
data = list(registrations_kinderlager.find())
data_xxl = list(registrations_xxl.find())
```

```python
# Create a DataFrame from the list of dictionaries
df_kinderlager = pd.DataFrame(data)
df_kinderlager = df_kinderlager.drop(columns=['_id'])

df_xxl = pd.DataFrame(data_xxl)
df_xxl = df_xxl.drop(columns=['_id'])
#df_kinderlager["birthday"] = df_kinderlager["birthday"].apply(lambda x: datetime.strptime(x, '%Y-%m-%d').date())
#print(df_kinderlager.columns)
#df_kinderlager
```

```python
df_kinderlager
```

```python
df_kinderlager["city"] = df_kinderlager["city"].str.strip()
```

```python
queryNotDoeggingen ='and city != "Doeggingen" and memberOf != "Ministranten Doeggingen"'
m = df_kinderlager.query('sexe == "m"')
md = df_kinderlager.query('sexe == "m" and (city == "Doeggingen" or memberOf == "Ministranten Doeggingen")')
md["group"] = "md"
m1 = df_kinderlager.query('sexe == "m" and birthday > "2015-07-29" ' + queryNotDoeggingen)
m1["group"] = "m1"
m2 = df_kinderlager.query('sexe == "m" and birthday < "2015-07-30" and birthday > "2014-07-29" ' + queryNotDoeggingen)
m2["group"] = "m2"
m3 = df_kinderlager.query('sexe == "m" and birthday < "2014-07-30" and birthday > "2013-07-29" ' + queryNotDoeggingen)
m3["group"] = "m3"
m4 = df_kinderlager.query('sexe == "m" and birthday < "2013-07-30" ' + queryNotDoeggingen)
m4["group"] = "m4"

w = df_kinderlager.query('sexe == "w"')
wd = df_kinderlager.query('sexe == "w" and (city == "Doeggingen" or memberOf == "Ministranten Doeggingen")')
wd["group"] = "wd"
w1 = df_kinderlager.query('sexe == "w" and birthday > "2015-07-29" ' + queryNotDoeggingen)
w1["group"] = "w1"
w2 = df_kinderlager.query('sexe == "w" and birthday < "2015-07-30" and birthday > "2014-07-29" ' + queryNotDoeggingen)
w2["group"] = "w2"
w3 = df_kinderlager.query('sexe == "w" and birthday < "2014-07-30" and birthday > "2013-07-29" ' + queryNotDoeggingen)
w3["group"] = "w3"
w4 = df_kinderlager.query('sexe == "w" and birthday < "2013-07-30" ' + queryNotDoeggingen)
w4["group"] = "w4"

print("Kinder insgesamt:", len(df_kinderlager))
print("-------------------------------")
print("Jungs insgesamt:", len(m))
print("Dögginger Jungs:", len(md))
print("M1 (? und ?):", len(m1))
print("M2 (? und ?):", len(m2))
print("M3 (Eric und Hannes):", len(m3))
print("M4 (Jari und Luis):", len(m4))
print("-------------------------------")
print("Mädels insgesamt:", len(w))
print("Dögginger Mädels:", len(wd))
print("W1 (? und ?):", len(w1))
print("W2 (? und ?):", len(w2))
print("W3 (Jule und Marie):", len(w3))
print("W4 (Maria und Maja):", len(w4))
```
